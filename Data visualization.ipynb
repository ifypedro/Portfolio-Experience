```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
df_used_cars = pd.read_csv("C:/Users/khay/Downloads/bar_chart_data.csv")
df_used_cars 
```

```python
### BAR CHART

plt.figure(figsize = (5,2))
plt.bar(x = df_used_cars['Brand'],
         height = df_used_cars['Cars Listings'], color = ['r','g','b','w','y','m','c'])
plt.xticks(rotation = 45, fontsize = 13)
plt.title("Cars Listing by Brands", fontsize = 16, fontweight = "bold")
plt.ylabel("Number of Listings", fontsize = 13)
plt.xlabel("Car Brands", fontsize = 13)
plt.yticks(rotation = 0)
plt.savefig("used cars bar.png")
plt.legend(df_used_cars)
plt.show()
```

```python
### PIE CHART

df_fuel_engine_type = pd.read_csv("pie_chart_data.csv")
df_fuel_engine_type
```

```python
plt.figure(figsize = (5,8))
plt.pie(df_fuel_engine_type['Number of Cars'],
         labels = df_fuel_engine_type['Engine Fuel Type'],
         autopct = '%.2f%%',
         textprops = {'size':'large','fontweight':'bold','rotation': 45,'color':'w'})


plt.legend()
plt.title('Cars by Engine Fuel Type', fontsize = 15, fontweight = 'bold')
plt.savefig('Cars by Engine Fuel Type.png')
plt.show()
```

```python
# Area Chart (stacked chart)

df_fuel_engine_types = pd.read_csv("stacked_area_chart_data.csv")
df_fuel_engine_types.head()
```

```python
sns.set_style("white")
colors = ["#133C55", "#386FA4", "#59A5D8",]
labels = ["Gas", "Petrol", "Diesel"]
plt.figure(figsize = (12,6))
plt.stackplot(df_fuel_engine_types['Year'],
              df_fuel_engine_types['Gas'],
              df_fuel_engine_types['Petrol'],
              df_fuel_engine_types['Diesel'], 
              colors = colors,
              edgecolor = 'none')

plt.xticks(df_fuel_engine_types['Year'], rotation = 45)
plt.ylabel("Number of Cars", fontsize = 13)
plt.xlabel("Years (1982 - 2016)", fontsize = 13)
plt.legend(labels = labels, loc = "upper left")
plt.title('Population of Engine Fuel Types (1982 - 2016)', fontsize = 14, weight = 'bold')
sns.despine()
#plt.savefig('second Population of Engine Fuel Types (1982 - 2016).png')
plt.show()
```

```python
#LINE CHARTS

df_spx_ftse_00_10 = pd.read_csv("line_chart_data.csv")
mod_df_spx_ftse_00_10  = df_spx_ftse_00_10.copy()
mod_df_spx_ftse_00_10.head(2)
```

```python
mod_df_spx_ftse_00_10['New_Date'] = pd.to_datetime(mod_df_spx_ftse_00_10['Date'])
new_mod_df_spx_ftse_00_10 = mod_df_spx_ftse_00_10
new_mod_df_spx_ftse_00_10.head(2)
```

```python
new_mod_df_spx_ftse_00_10 = new_mod_df_spx_ftse_00_10.drop(['Date'], axis = 1)
new_mod_df_spx_ftse_00_10.head(2)
```

```python
#LINE CHART

labels = ['S&P 500', 'FTSE 100']
plt.figure(figsize = (20 , 8))
plt.plot(new_mod_df_spx_ftse_00_10['New_Date'], new_mod_df_spx_ftse_00_10['GSPC500'])
plt.plot(new_mod_df_spx_ftse_00_10['New_Date'], new_mod_df_spx_ftse_00_10['FTSE100'])
plt.title("S&P vs FTSE Return (2000 - 2010)", fontsize = 14, fontweight = 'bold')
plt.xlabel("Returns)", fontsize = 13)
plt.ylabel("Date)", fontsize = 13)
plt.legend(labels = labels, fontsize = 'large')
#plt.savefig('S&P vs FTSE Return (2000 - 2010).png')
plt.show()
```

```python
new_mod_df_spx_ftse_H2_08 = new_mod_df_spx_ftse_00_10[(new_mod_df_spx_ftse_00_10.New_Date >= '2008-07-01') &
                                                       (new_mod_df_spx_ftse_00_10.New_Date <= '2008-12-03')]

new_mod_df_spx_ftse_H2_08.head()
```

```python
#labels = ['S&P 500', 'FTSE 100']
plt.figure(figsize = (20 , 8))
plt.plot(new_mod_df_spx_ftse_H2_08['New_Date'], new_mod_df_spx_ftse_H2_08['GSPC500'], color = 'midnightblue')
plt.plot(new_mod_df_spx_ftse_H2_08['New_Date'], new_mod_df_spx_ftse_H2_08['FTSE100'], color = 'crimson')
plt.title("S&P vs FTSE Return (H2 2008)", fontsize = 14, fontweight = 'bold')
plt.xlabel("Returns)", fontsize = 13)
plt.ylabel("Date)", fontsize = 13)
plt.legend(labels = labels,fontsize = 'large')
#plt.savefig('S&P vs FTSE Return (H2 2008).png')
plt.show()
```

```python
#HISTOGRAM

df_real_estate = pd.read_csv("histogram_data.csv")
df_real_estate.head()
```

```python
sns.set_style('white')
plt.figure(figsize = (8, 6))
plt.hist(df_real_estate['Price'], bins = 8,
         color = '#108A99')
plt.title('Distribution of Real Prices', fontsize = 14, weight = 'bold')
plt.ylabel('Number of Properties')
sns.despine()
#plt.savefig('Distribution of Real Prices.png')
plt.show()
```

```python
df_real_estate_scatter = pd.read_csv("scatter_data.csv")
df_real_estate_scatter.head()
```

```python
plt.figure(figsize = (14, 5))
scatter = plt.scatter(df_real_estate_scatter['Area (ft.)'], 
            df_real_estate_scatter['Price'],
            alpha = 0.6,
            c = df_real_estate_scatter['Building Type'],
            cmap = 'viridis')

plt.legend(*scatter.legend_elements(),
          loc = 'upper left',
          title = 'Building Type')
plt.title('Relationship between Area and Price of Califonia Real Estate', fontsize = 14, weight = 'bold')
plt.xlabel('Area (ft.)',  weight = 'bold')
plt.ylabel("Price (000's of $)")
#sns.despine()
#plt.savefig('Relationship between Area and Price of Califonia Real Estate(matlab).png')
plt.show()
```

```python
plt.figure(figsize=(14, 5))
sns.scatterplot(data=df_real_estate_scatter,
                x='Area (ft.)', 
                y='Price',
                hue='Building Type',
                palette=['black', 'darkblue', 'purple', 'pink', 'lightgray'],  # Changed white to lightgray
                s=100)
plt.title('Relationship between Area and Price of California Real Estate', fontsize=14, weight='bold')
plt.xlabel('Area (ft.)', weight='bold')
plt.ylabel("Price (000's of $)", weight='bold')
#plt.savefig('Relationship between Area and Price of Califonia Real Estate(seaborn).png')
plt.show()
```

```python
df_ad_expenditure = pd.read_csv("scatter_plot_ii.csv")
df_ad_expenditure.head()
```

```python
sns.set(rc = {'figure.figsize': (14,5)})
sns.regplot(x = "Budget",
            y = "Sales",
            data = df_ad_expenditure,
            scatter_kws = {'color' : 'k'},
            line_kws = { 'color': 'red'})

plt.title('Budget against Sale', fontsize=14, weight='bold')
plt.xlabel('Ad Expenditures', weight='bold')
plt.ylabel("Ad Sales", weight='bold')
#plt.savefig('Budget against Sales (sns.regplot).png')
plt.show()
```

```python
sns.set(rc = {'figure.figsize': (10,5)})
sns.lmplot(x = "Budget",
            y = "Sales",
            data = df_ad_expenditure,
            height = 5,
            scatter_kws = {'color' : 'k'},
            line_kws = { 'color': 'red'})

plt.title('Budget against Sale', fontsize=14, weight='bold')
plt.xlabel('Ad Expenditures', weight='bold')
plt.ylabel("Ad Sales", weight='bold')
#plt.savefig('Budget against Sales (lmplot).png')
plt.show()
```

```python
from matplotlib.ticker import PercentFormatter
```

```python
df_kdnuggets = pd.read_csv('bar_line_chart_data.csv')
df_kdnuggets.head()
```

```python
fig, ax = plt.subplots(figsize = (10,7))
ax.bar(df_kdnuggets['Year'],
        df_kdnuggets['Participants'],
        color = ('k','b','y'))
ax.set_ylabel("Number of Participants",
                weight = 'bold')
ax.tick_params(axis = 'y',
                width = 2, 
                labelsize = 'large')
ax1 = ax.twinx()
ax1.set_ylim(0, 1)
ax1.yaxis.set_major_formatter(PercentFormatter(xmax = 1.0))
ax1.plot(df_kdnuggets['Year'], 
         df_kdnuggets['Python Users'],
          color = '#b60000',
          marker = 'D')

ax1.set_ylabel('Python Users',
                color = '#b60000', weight = 'bold')
ax1.tick_params(axis = 'y',
                color = '#b60000',
                width = 2,
                labelsize = 'large')
#plt.savefig("KD Nuggets Survey Python Users (2012 - 2019).png")
ax.set_title("KD Nuggets Survey Python Users (2012 - 2019)", fontsize = 14, weight = 'bold')
```
